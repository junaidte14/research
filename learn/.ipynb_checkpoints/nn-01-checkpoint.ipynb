### 1. The Forward Pass Equations
For a single neuron, the computations happen in three steps:
1. **Linear combination:** 
   $$z = w_1 x_1 + w_2 x_2 + b$$
2. **Sigmoid activation function ($f(z)$):** 
   $$a = \frac{1}{1 + e^{-z}}$$
3. **Binary Cross-Entropy Loss:** 
   $$L = -\left(y \ln(a) + (1-y) \ln(1-a)\right)$$

### 2. Backpropagation & Chain Rule
We want to find how much each weight ($w_1, w_2$) and bias ($b$) affects the loss $L$. Using the chain rule:

$$\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w_1}$$

Step-by-step:
* **Loss w.r.t activation ($a$):** 
  $$\frac{\partial L}{\partial a} = \frac{a - y}{a(1-a)}$$
* **Activation w.r.t $z$ (Derivative of Sigmoid):** 
  $$\frac{\partial a}{\partial z} = a(1-a)$$
* **Combining them ($\frac{\partial L}{\partial z}$):** 
  $$\frac{\partial L}{\partial z} = \left(\frac{a - y}{a(1-a)}\right) \cdot (a(1-a)) = a - y$$
* **Final gradient for $w_1$:** 
  $$\frac{\partial L}{\partial w_1} = (a - y)x_1$$

### Complete Mathematical Derivation of Backpropagation

To understand how our single-neuron model learns, we use calculus and the **Chain Rule** to compute how much each weight ($w_1, w_2$) and bias ($b$) contributes to the final Binary Cross-Entropy Loss ($L$).

We want to find the partial derivative of the loss with respect to weight $w_1$:
$$\frac{\partial L}{\partial w_1} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w_1}$$

Let's derive each of these three pieces step-by-step.

---

#### Step 1: Derivative of Loss w.r.t. Activation ($a$)
**The Goal:** Find how a change in our predicted probability $a$ affects the Binary Cross-Entropy Loss $L$.

* **Starting Equation:**
  $$L = -y \ln(a) - (1-y) \ln(1-a)$$

* **Differentiation:**
  Differentiating $L$ with respect to $a$ (treating target $y$ as a constant):
  $$\frac{\partial L}{\partial a} = -y \left(\frac{1}{a}\right) - (1-y) \left(\frac{1}{1-a}\right) \cdot (-1)$$
  $$\frac{\partial L}{\partial a} = -\frac{y}{a} + \frac{1-y}{1-a}$$

* **Algebraic Simplification:**
  Find a common denominator $a(1-a)$:
  $$\frac{\partial L}{\partial a} = \frac{-y(1-a) + (1-y)a}{a(1-a)} = \frac{-y + ya + a - ya}{a(1-a)}$$
  
  The $+ya$ and $-ya$ terms cancel out, leaving:
  $$\frac{\partial L}{\partial a} = \frac{a - y}{a(1-a)}$$

---

#### Step 2: Derivative of Activation w.r.t. $z$ (Derivative of Sigmoid)
**The Goal:** Find how a change in the linear output $z$ affects the sigmoid activation output $a$.

* **Starting Equation:**
  $$a = \sigma(z) = \frac{1}{1 + e^{-z}} = (1 + e^{-z})^{-1}$$

* **Differentiation (Power Rule & Chain Rule):**
  $$\frac{\partial a}{\partial z} = -1 \cdot (1 + e^{-z})^{-2} \cdot (-e^{-z}) = \frac{e^{-z}}{(1 + e^{-z})^2}$$

* **Algebraic Simplification into $a$:**
  Split the fraction and add/subtract 1 in the numerator:
  $$\frac{\partial a}{\partial z} = \left(\frac{1}{1 + e^{-z}}\right) \cdot \left(\frac{e^{-z}}{1 + e^{-z}}\right)$$
  $$\frac{\partial a}{\partial z} = a \cdot \left(\frac{(1 + e^{-z}) - 1}{1 + e^{-z}}\right) = a \cdot (1 - a)$$

---

#### Step 3: Derivative of $z$ w.r.t. Parameters
**The Goal:** Find how changes in weights and bias affect the linear combination $z = w_1 x_1 + w_2 x_2 + b$.

* **Partial Derivatives:**
  * $\frac{\partial z}{\partial w_1} = x_1$
  * $\frac{\partial z}{\partial w_2} = x_2$
  * $\frac{\partial z}{\partial b} = 1$

---

#### Step 4: Putting It All Together (The Chain Rule)
Now, we multiply all the pieces together using the chain rule:

$$\frac{\partial L}{\partial w_1} = \underbrace{\frac{a - y}{a(1-a)}}_{\frac{\partial L}{\partial a}} \cdot \underbrace{a(1-a)}_{\frac{\partial a}{\partial z}} \cdot \underbrace{x_1}_{\frac{\partial z}{\partial w_1}}$$

Notice how the complex fraction terms $\frac{1}{a(1-a)}$ and $a(1-a)$ completely cancel each other out! This leaves us with an elegantly simple final gradient:

$$\frac{\partial L}{\partial w_1} = (a - y) x_1$$
$$\frac{\partial L}{\partial w_2} = (a - y) x_2$$
$$\frac{\partial L}{\partial b} = (a - y)$$

In our Python code, we define `dz = a - y` to represent this shared error term, making gradient updates exceptionally fast and clean!

```python
import math
learning_rate = 0.5
epochs = 200

points = [(0,0), (0,1), (1,0), (1,1)]
targets = [0, 0, 0, 1] # For AND Operation
dataset = list(zip(points, targets))
print(dataset)

#starting values
w1 = 0.5
w2 = -0.3
b = 0.0
```

```python
# training loop
losses = list()
for epoch in range(1, epochs+1):
    #print("=" * 74)
    #print(f"Epoch: {epoch}, w1: {w1}, w2: {w2}, b: {b}")
    #print("=" * 74)

    sum_dw1 = 0.0
    sum_dw2 = 0.0
    sum_db = 0.0
    sum_loss = 0.0

    for (x1,x2), y in zip(points, targets):
        #print("-" * 74)
        #forward pass
        z = w1*x1 + w2*x2 + b
        #print(f"z = w1*x1 + w2*x2 + b : {z}")
        a = 1 / (1 + math.exp(-z)) # sigmoid: used to contain the z to 0 to 1 instead of -$-\infty$ to +$-\infty$
        #print(f"math.exp(-z): {math.exp(-z)}")
        #print(f"a = 1 / (1 + math.exp(-z)): {a}")
        loss = -(y * math.log(a) + (1-y) * math.log(1-a))
        #print(f"math.log(a): {math.log(a)}")
        #print(f"math.log(1-a): {math.log(1-a)}")
        #print(f"a: {a} and (1-a): {1-a}")
        #print(f"loss = -(y * math.log(a) + (1-y) * math.log(1-a)): {loss}")
        #print("-" * 74)

        #backward pass (backpropagation) - (gradients for this one point)
        dz = a - y
        dw1 = dz * x1
        dw2 = dz * x2
        db = dz

        #print(f"point: ({x1}, {x2}), target label is: {y}")
        #print(f"backward: dz = a - y = {dz},   dw1 = {dw1}"
              #f"   dw2 = {dw2}   db = {db}")

        sum_dw1 += dw1
        sum_dw2 += dw2
        sum_db += db
        sum_loss += loss

    n = len(points)
    g_w1 = sum_dw1/n #average gradient of w1
    g_w2 = sum_dw2/n #average gradient of w2
    g_b = sum_db/n #average gradient of b

    #print("-" * 74)
    #print(f"mean loss = {sum_loss / n}")
    #print(f"mean gradients: dw1 = {g_w1}   dw2 = {g_w2}   db = {g_b}")
    losses.append(sum_loss/n)
    
    # gradient descent update
    #print(f"update: w1 = {w1} - {learning_rate}*({g_w1}) = {w1 - learning_rate * g_w1}")
    #print(f"        w2 = {w2} - {learning_rate}*({g_w2}) = {w2 - learning_rate * g_w2}")
    #print(f"        b  = {b} - {learning_rate}*({g_b}) = {b - learning_rate * g_b}")
    w1 -= learning_rate * g_w1
    w2 -= learning_rate * g_w2
    b -= learning_rate * g_b

#print("=" * 74)
#print(f"final: w1={w1:+.4f}  w2={w2:+.4f}  b={b:+.4f}")
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1, epochs+1), losses, marker="o")
plt.title("Training Loss Over Epochs")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.grid(True)
plt.draw()
```

```python
# --- Accuracy Evaluation ---
correct_predictions = 0
total_points = len(points)

print("=" * 74)
print("Final Model Evaluation & Accuracy")
print("=" * 74)

for (x1, x2), y in zip(points, targets):
    # Forward pass using the final trained weights
    z = w1 * x1 + w2 * x2 + b
    a = 1 / (1 + math.exp(-z))
    
    # Threshold probability at 0.5 to get binary prediction (0 or 1)
    prediction = 1 if a >= 0.5 else 0
    
    # Check if prediction matches the target
    is_correct = (prediction == y)
    if is_correct:
        correct_predictions += 1
        
    print(f"Point: ({x1}, {x2}) | Target: {y} | Prob (a): {a:.4f} | Pred: {prediction} | Correct: {is_correct}")

accuracy = (correct_predictions / total_points) * 100
print("-" * 74)
print(f"Final Accuracy: {accuracy}% ({correct_predictions}/{total_points} correct)")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Separate points based on their target labels for plotting
x_zeros = [p[0] for p, t in zip(points, targets) if t == 0]
y_zeros = [p[1] for p, t in zip(points, targets) if t == 0]
x_ones  = [p[0] for p, t in zip(points, targets) if t == 1]
y_ones  = [p[1] for p, t in zip(points, targets) if t == 1]

# Set up the plot
plt.figure(figsize=(6, 6))
plt.scatter(x_zeros, y_zeros, color='red', s=120, label='Target 0 (False)', zorder=3)
plt.scatter(x_ones, y_ones, color='blue', s=120, label='Target 1 (True)', zorder=3)

# Calculate the decision boundary line: x2 = (-w1/w2)*x1 - (b/w2)
x1_vals = np.linspace(-0.5, 1.5, 100)
if w2 != 0:
    x2_vals = -(w1 * x1_vals + b) / w2
    plt.plot(x1_vals, x2_vals, color='green', linestyle='--', linewidth=2.5, label='Decision Boundary', zorder=2)

# Graph formatting
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.axhline(0, color='gray', linewidth=1, linestyle=':')
plt.axvline(0, color='gray', linewidth=1, linestyle=':')
plt.xlabel('$x_1$', fontsize=12)
plt.ylabel('$x_2$', fontsize=12)
plt.title('Single Neuron Decision Boundary (AND Gate)', fontsize=12, fontweight='bold')
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import math

def draw_equation_neuron(w1, w2, b, sample_point=(1, 1)):
    x1, x2 = sample_point
    
    # Calculate intermediate values for the sample point
    z = w1 * x1 + w2 * x2 + b
    a = 1 / (1 + math.exp(-z))
    
    fig, ax = plt.subplots(figsize=(12, 6.5))
    ax.set_aspect('equal')
    ax.axis('off')
    
    # Node Coordinates
    pos_x1 = (1, 4.8)
    pos_x2 = (1, 1.7)
    pos_z  = (4.8, 3.25)     # Summation node (z)
    pos_a  = (8.2, 3.25)     # Activation node (a)
    pos_out= (11, 3.25)      # Output node
    
    # --- 1. Draw Connections & Weights ---
    # x1 to z
    ax.annotate("", xy=pos_z, xytext=pos_x1,
                arrowprops=dict(arrowstyle="->", color="royalblue", lw=2.5, shrinkA=22, shrinkB=28))
    ax.text(2.5, 4.3, f"$w_1$ = {w1:.3f}", fontsize=11, color="darkblue", fontweight='bold', backgroundcolor='white')
    
    # x2 to z
    ax.annotate("", xy=pos_z, xytext=pos_x2,
                arrowprops=dict(arrowstyle="->", color="royalblue", lw=2.5, shrinkA=22, shrinkB=28))
    ax.text(2.5, 2.1, f"$w_2$ = {w2:.3f}", fontsize=11, color="darkblue", fontweight='bold', backgroundcolor='white')
    
    # z to a
    ax.annotate("", xy=pos_a, xytext=pos_z,
                arrowprops=dict(arrowstyle="->", color="darkorange", lw=2.5, shrinkA=28, shrinkB=28))
    ax.text(6.2, 3.5, "Passes $z$", fontsize=10, color="darkorange", fontweight='bold')
    
    # a to Output
    ax.annotate("", xy=pos_out, xytext=pos_a,
                arrowprops=dict(arrowstyle="->", color="forestgreen", lw=2.5, shrinkA=28, shrinkB=15))
    ax.text(9.4, 3.5, "Threshold (0.5)", fontsize=10, color="forestgreen", fontweight='bold')

    # --- 2. Draw Nodes with Equations ---
    
    # Input Node 1 ($x_1$)
    ax.add_patch(plt.Circle(pos_x1, 0.55, color='skyblue', ec='black', lw=2, zorder=3))
    ax.text(pos_x1[0], pos_x1[1]+0.2, "$x_1$", fontsize=12, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_x1[0], pos_x1[1]-0.1, f"Val: {x1}", fontsize=10, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_x1[0], pos_x1[1]-0.32, "Input Feature", fontsize=8, ha='center', va='center', style='italic', zorder=4)

    # Input Node 2 ($x_2$)
    ax.add_patch(plt.Circle(pos_x2, 0.55, color='skyblue', ec='black', lw=2, zorder=3))
    ax.text(pos_x2[0], pos_x2[1]+0.2, "$x_2$", fontsize=12, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_x2[0], pos_x2[1]-0.1, f"Val: {x2}", fontsize=10, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_x2[0], pos_x2[1]-0.32, "Input Feature", fontsize=8, ha='center', va='center', style='italic', zorder=4)

    # Summation Node (z) - Equation: z = w1*x1 + w2*x2 + b
    ax.add_patch(plt.Circle(pos_z, 0.95, color='gold', ec='black', lw=2, zorder=3))
    ax.text(pos_z[0], pos_z[1]+0.5, r"$\sum$ (Linear Step)", fontsize=10, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_z[0], pos_z[1]+0.15, r"$z = \sum (w_i x_i) + b$", fontsize=9, ha='center', va='center', zorder=4)
    ax.text(pos_z[0], pos_z[1]-0.2, f"= ({w1:.2f}×{x1}) + ({w2:.2f}×{x2}) + {b:.2f}", fontsize=8, ha='center', va='center', zorder=4)
    ax.text(pos_z[0], pos_z[1]-0.5, r"$\mathbf{z = " + f"{z:.2f}" + "}$", fontsize=10, ha='center', va='center', color='darkred', fontweight='bold', zorder=4)

    # Activation Node (a) - Equation: a = 1 / (1 + e^-z)
    ax.add_patch(plt.Circle(pos_a, 0.95, color='orange', ec='black', lw=2, zorder=3))
    ax.text(pos_a[0], pos_a[1]+0.5, "Sigmoid (Activation)", fontsize=10, ha='center', va='center', fontweight='bold', zorder=4)
    ax.text(pos_a[0], pos_a[1]+0.15, r"$a = \frac{1}{1 + e^{-z}}$", fontsize=9, ha='center', va='center', zorder=4)
    ax.text(pos_a[0], pos_a[1]-0.2, f"= 1 / (1 + e^-({z:.2f}))", fontsize=8, ha='center', va='center', zorder=4)
    ax.text(pos_a[0], pos_a[1]-0.5, r"$\mathbf{a = " + f"{a:.3f}" + "}$", fontsize=10, ha='center', va='center', color='darkred', fontweight='bold', zorder=4)

    # Output Box
    bbox_props = dict(boxstyle="round,pad=0.6", fc="lightgreen", ec="black", lw=2)
    pred_val = 1 if a >= 0.5 else 0
    ax.text(pos_out[0], pos_out[1], f"Prediction\n\nPred: {pred_val}\n({a*100:.1f}%)", fontsize=10, ha='center', va='center', fontweight='bold', bbox=bbox_props, zorder=4)

    # Title & Formatting
    plt.xlim(0, 12.5)
    plt.ylim(0.2, 5.8)
    plt.title(f"Neuron Mathematical Trace for Input Sample: ({x1}, {x2})", fontsize=14, fontweight='bold', pad=20)
    plt.show()
```
